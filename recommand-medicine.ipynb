```python
from sklearn.feature_extraction.text import TfidfVectorizer

tdf_vector = TfidfVectorizer() 
tdf_vector_matrix = tdf_vector.fit_transform(data["combined_features"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity


def recommend_cheapest(name, top_n=5):
    idx = data[data['name'] == name].index[0]
    
    sim_scores = cosine_similarity(tdf_vector_matrix[idx], tdf_vector_matrix).flatten()
    
    similar_indices = sim_scores.argsort()[-20:]  # take more candidates
    
    result = data.iloc[similar_indices]
    
    # return result.sort_values('price(₹)').head(top_n)
    return result["Consolidated_Side_Effects"]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_recommendations_safe(medicine_name, data, tdf_vector_matrix, tfidf_vectorizer):
    # 1. User ki search ki hui medicine ka index dhoondein
    idx = data[data['name'].str.contains(medicine_name, case=False, na=False)].index
    
    if len(idx) == 0:
        return "Medicine not found."
    
    # Hum sirf pehli matching medicine uthayenge
    target_idx = idx[0]
    
    # 2. POORE matrix ki jagah sirf IS ROW ki similarity nikaalein
    # tfidf_matrix[target_idx] sirf ek row hai
    cosine_sim_row = cosine_similarity(tdf_vector_matrix[target_idx], tdf_vector_matrix).flatten()
    
    # 3. Baaki logic same rahega
    sim_scores = list(enumerate(cosine_sim_row))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    
    # Top 10 matches (excluding itself)
    recommendations = []
    current_med_price = data.iloc[target_idx]['price(₹)']
    
    for i, score in sim_scores[1:50]: # Top 50 potential matches
        if data.iloc[i]['price(₹)'] < current_med_price:
            recommendations.append(data.iloc[i])
        if len(recommendations) >= 10: # Top 5 sasti dawaiyan milte hi ruk jayein
            break
            
    return pd.DataFrame(recommendations)

```

```python
result = get_recommendations_safe("Allegra 120mg", data, tdf_vector_matrix, tdf_vector)
print(result[['name', 'price(₹)']])
```

```python
def get_scanned_medicine_details(medicine_name, data):
    # Partial match se dawai dhoondein (Safe approach)
    match = data[data['name'].str.contains(medicine_name, case=False, na=False)]
    
    if match.empty:
        return "Sorry, ye dawai hamare database mein nahi hai."
    
    # Pehli matching dawai ki details nikal lein
    med_info = match.iloc[0]
    
    # Ek dictionary return karein jise frontend (Streamlit) easily padh sake
    return {
        "Name": med_info['name'],
        "Side_Effects": med_info['Consolidated_Side_Effects'],
        "Benefits":med_info['combined_features']
        # Aap chahein toh yahan 'Uses' ya 'Composition' bhi add kar sakte hain
    }

# Test karein:
details = get_scanned_medicine_details("Povinovam Ointment", data)
print(details["Name"])
print(details["Side_Effects"])
print(details["Benefits"])
```
